```python
#Trentin Harris
#Academic Award Tracker Program
#This Program collects students first and last names, along with their GPA then tests if they made Dean's list or honor role
student_info = {
    
}

student_last = input(print("Please enter student's last name: "))


while student_last != "ZZZ":
    student_first = input(print("Please enter student's first name: "))
    student_name = student_first + " " + student_last
    
    gpa = float(input(print("Please enter student's GPA: ")))
    student_info['student_name'] = gpa
    
    if gpa >= 3.5:
        print(student_name, "has made the deans list!")
    elif gpa >= 3.25:
        print(student_name, "has made Honor Roll!")
    else:
        break

    student_last = input(print("Please enter student's last name or 'ZZZ' to quit: "))
```
