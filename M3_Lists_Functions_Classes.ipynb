```python
# Superclass
class Vehicle:
    def __init__(self, Vehicle_Type):
        self.Vehicle_Type = Vehicle_Type

# Subclass 
class Automobile(Vehicle):
    def __init__(self, Vehicle_Type, Year, Make, Model, Num_Door, Type_Roof):
        super().__init__(Vehicle_Type)

        self.Year = Year
        self.Make = Make
        self.Model = Model
        self.Num_Door = Num_Door
        self.Type_Roof = Type_Roof
    
# Main
Vehicle_Type = input("Enter Vehicle Type: ")

Year = input("Year: ")
Make = input("Make: ")
Model = input("Model: ")
Num_Door = input("Number of doors: ")
Type_Roof = input("Type of roof: ")

user_car = Automobile(
    Vehicle_Type,
    Year,
    Make,
    Model,
    Num_Door,
    Type_Roof
)

print("\nVehicle Info")
print("Vehicle type: ", user_car.Vechicle_Type)
print("Year: ", user_car.Year)
print("Make: ", user_car.Make)
print("Model: ", user_car.Model)
print("Number of doors: ", user_car.Num_Door)
print("Type of roof: ", user_car.Type_Roof)
```
