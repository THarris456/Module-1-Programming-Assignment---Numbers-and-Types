```python
class Solution:
    def sort012(self, arr):
        # code here
        arr.sort()
        return arr

#Main
arr = [0, 1, 2, 0, 1, 2]
print(arr)

s = Solution()
r = s.sort012(arr)

print(r)
```

```python
class Solution:
    def firstSearch(self, arr, k):
        # Code Here
        ind = list(filter(lambda i: arr[i] == k, range(len(arr))))
        return ind[0] if ind else -1

#Main
arr = [1, 2, 2, 2, 3]
k = 2

print(arr)

s = Solution()
print(s.firstSearch(arr, k))
```
