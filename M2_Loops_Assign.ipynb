```python
guess_me = 7
number = 1

while number != guess_me:
    if number < guess_me:
        print(number, "too low")
        number = number + 1
    else:
        print(number, "oops")
        number = number - 1
print("found it")
```

```python
guess_me = 5

for number in range(10):
    if number < guess_me:
        print("too low")
    elif number > guess_me:
        print("oops")
        break    #not reachable but would stop unnecessary iteration. 
    else:
        print("found it")
        break
```
